```python
class Machine:
    def __init__(self):
        self.op_ready = []
        self.op_by_order = [[] for _ in range(6)]
        self.current_position = 0
        

class Job:
    def __init__(self):
        self.op = []
        self.availability = [True] + [False for _ in range(6 -1)]
        

machine_list = [Machine() for i in range(6)]
job_list = [Job() for i in range(6)]
for i in range(6):
    for j in range(6):
        # machine index = op_data[i][j][0]
        # current operation order = j
        machine_list[op_data[i][j][0]].op_by_order[j].append(i)
    
    while len(machine_list[i].op_by_order[machine_list[i].current_position]) == 0:
        machine_list[i].current_position += 1
    machine_list[i].op_ready = machine_list[i].op_by_order[machine_list[i].current_position]
```

```python
class Operation:
    def __init__(self, i, j, k):
        self.job = i
        self.precedence = j
        self.machine = k
        
        
        if j == 0:
            self.job_ready, self.machine_ready = True
        else:
            self.job_ready, self.machine_ready = False
            
        self.op_prior = None
        self.op_following = None
        
class JSSP:
    def __init__(self, op_data):
        self.op_data = op_data
        self.op_list = [[] for _ in range(6)]
        self.machine_list = [Machine() for _ in range(6)]

        
        # Initialization
        for i in range(6):
            for j in range(6):
                self.op_list[i].append(Operation(i, j, op_data[i][j][0]))
        
        for i in range(6):
            for j in range(6):
                self.machine_list[op_data[i][j][0]].op_by_order[j].append(self.op_list[i][j])
            while len(self.machine_list[i].op_by_order[self.machine_list[i].current_position]) == 0:
                self.machine_list[i].current_position += 1
            self.machine_list[i].op_ready = self.machine_list[i].op_by_order[machine_list[i].current_position]
            for op in self.machine_list[i].op_ready:
                op.machine_ready = True
                
        
        # 연결 관계 수립
        for i in range(6):
            for j in range(1, 6):
                self.op_list[i][j].op_prior = self.op_list[i][j-1]
                self.op_list[i][j-1].op_following = self.op_list[i][j]
        
        
        
```
